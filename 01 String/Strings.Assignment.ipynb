```python
s = input("Enter string: ")
p = int(input("Enter position: "))

new = ''
for i in range(len(s)):
    if i!=p:
        new += s[i]

print(new)
```

```python
s = input("Enter string: ").lower()
vowels = "aeiou"

for v in vowels:
    count = 0
    for ch in s:
        if ch == v:
            count += 1
    print(v, "=", count)
```

```python
s = input("Enter string: ").lower()
v = 'aeiou'
result = ""

for ch in s:
    if ch not in v:
        result += ch
print(result)
```

```python
s = input("Enter Full Name: ")
print(s[0], end=".")

for i in range(len(s)):
    if s[i] == " ":
        print(s[i+1], end=".")
        last = i+1

print(s[last:])
```
