```python
marks = np.array([ 

    [78, 85, 90, 88, 76], 

    [56, 65, 70, 72, 60], 

    [91, 95, 89, 97, 93], 

    [45, 50, 40, 55, 48], 

    [68, 72, 75, 70, 69], 

    [88, 90, 92, 91, 89], 

    [35, 42, 38, 40, 45], 

    [80, 78, 82, 85, 79] 

]) 
```

```python
print(marks)
```

```python
print("Shape :", marks.shape)
print("Size :", marks.size)
print("Dimensions :", marks.ndim)
print("Datatype :", marks.dtype)
```

```python
total_marks = np.sum(marks, axis=1)
print("\nTotal Marks:")
print(total_marks)
```

```python
average_marks = np.mean(marks, axis = 1)
print('\nAverage Marks:')
print(average_marks)
```

```python
subject_average = np.mean(marks, axis=0)
print("\nSubject-wise Average:")
print(subject_average)
```

```python
highest_marks = np.max(marks, axis=0)
print("\nHighest Marks:")
print(highest_marks)
```

```python
lowest_marks = np.min(marks, axis=0)
print("\nLowest Marks:")
print(lowest_marks)
```

```python
print(max(total_marks))
```

```python
top_student = np.argmax(total_marks)

print("\nTop Student Index :", top_student)
print("Total Marks :", total_marks[top_student])
```
